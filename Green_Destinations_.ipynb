#Question 1: What is the attrition rate?

```python
# Python calculation of attrition rate
attrition_rate = df['Attrition'].value_counts(normalize=True) * 100
attrition_rate
```

# Question 2: Does age affect attrition?

In Tableau:
- Plot bar chart of Age (binned) vs. Attrition

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(data=df, x='Age', hue='Attrition', multiple='stack', bins=10)
plt.title("Attrition by Age")
plt.show()
```
